**Carregando o arquivo**

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv(
    "amazon-tratado-v2.csv",
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)

pd.set_option("display.max_columns", None)

print(f"Registros: {df.shape[0]:,}")
print(f"Colunas: {df.shape[1]}")

display(df.head())
```

**Conhecendo as colunas e exemplos de dados**

```python
estrutura = pd.DataFrame({
    "coluna": df.columns,
    "tipo": df.dtypes.astype(str).values,
    "exemplo": [
        df[coluna].dropna().astype(str).iloc[0][:150]
        if not df[coluna].dropna().empty else ""
        for coluna in df.columns
    ]
})

display(estrutura)
```

**Explorando o portfólio disponível**

```python
portfolio = (
    df.groupby("categoria_nivel1")
      .agg(
          produtos=("product_id", "nunique"),
          marcas=("marca", "nunique")
      )
      .sort_values("produtos", ascending=False)
)

display(portfolio)

print("\nExemplos de produtos e sua classificação:")

display(
    df[[
        "product_name",
        "marca",
        "categoria_nivel1",
        "categoria_nivel2",
        "categoria_nivel3"
    ]].head(10)
)
```

**Observando as escalas de preço, desconto, nota e engajamento**

```python
indicadores = [
    "preco_original",
    "preco_com_desconto",
    "desconto_pct",
    "rating",
    "qtd_avaliacoes"
]

display(
    df[indicadores]
      .describe()
      .loc[["min", "25%", "50%", "75%", "max"]]
      .T
      .rename(columns={"50%": "mediana"})
      .round(2)
)

print("Em desconto_pct, 0.43 significa 43%.")
```

**Visualizando os campos comerciais juntos**

```python
display(
    df[[
        "product_name",
        "categoria_nivel1",
        "marca",
        "preco_original",
        "preco_com_desconto",
        "desconto_pct",
        "qtd_avaliacoes",
        "rating"
    ]].head(15)
)
```

**Explorando os textos para Customer Insights**

```python
amostra_textos = df[[
    "product_name",
    "rating",
    "about_product",
    "review_title",
    "review_content"
]].head(3)

for _, produto in amostra_textos.iterrows():
    print("=" * 80)
    print("PRODUTO:", produto["product_name"])
    print("NOTA DO PRODUTO:", produto["rating"])

    print("\nDESCRIÇÃO:")
    print(str(produto["about_product"])[:600])

    print("\nTÍTULOS DAS AVALIAÇÕES:")
    print(str(produto["review_title"])[:600])

    print("\nCONTEÚDO DAS AVALIAÇÕES:")
    print(str(produto["review_content"])[:1200])

    print()
```

**Relacionando os dados ao storytelling**

```python
storytelling = pd.DataFrame([
    {
        "tela": "1. Portfólio e preços",
        "pergunta": "O que compõe o catálogo e como os preços variam?",
        "campos": "product_id, marca, categoria_nivel1 a 7, preços",
        "indicadores": "Produtos distintos e mediana dos preços"
    },
    {
        "tela": "2. Descontos",
        "pergunta": "Onde os descontos são maiores?",
        "campos": "desconto_pct, categorias, preços",
        "indicadores": "Desconto mediano e produtos por faixa de desconto"
    },
    {
        "tela": "3. Engajamento",
        "pergunta": "Quais produtos recebem mais avaliações?",
        "campos": "product_id, qtd_avaliacoes, categorias, desconto_pct",
        "indicadores": "Ranking e mediana de avaliações por produto"
    },
    {
        "tela": "4. Satisfação",
        "pergunta": "Os produtos mais avaliados também têm boas notas?",
        "campos": "rating, qtd_avaliacoes, categorias",
        "indicadores": "Notas por categoria e matriz nota × avaliações"
    },
    {
        "tela": "5. Voz do cliente",
        "pergunta": "O que os comentários revelam sobre esses produtos?",
        "campos": "review_title, review_content, about_product",
        "indicadores": "Temas de elogios e reclamações por produto"
    }
])

display(storytelling)
```

**Sugestão de Storytelling**

Um caso de uso específico que combina com os dados é **priorizar produtos para campanhas promocionais considerando a satisfação dos clientes.**

O cenário seria: **a liderança precisa escolher quais produtos destacar na próxima campanha e quais precisam de investigação antes de receber mais promoção.**

A pergunta central pode ser:
**Quais produtos são candidatos a ganhar destaque promocional e quais apresentam sinais de insatisfação que merecem atenção primeiro?**

Os cinco itens que contribuiriam para essa decisão:

**1. Portfólio e preços:** Identificar categorias, marcas e faixas de preço para comparar produtos semelhantes
**2.Descontos:** Examinar os descontos atuais e sua associação com notas e quantidade de avaliações, verificar produtos caros mas que tem boas avaliações e avaliar a possibilidade de oferecer desconto nesses produtos.
**3. Engajamento:** Identificar produtos com muitas ou poucas avaliações dentro de cada categoria
**4.Satisfação:** Separar candidatos a destaque dos produtos com sinais de insatisfação
**5. Voz do cliente:** Investigar elogios e reclamações e justificar as recomendações finais

O resultado seria uma **lista de produtos com ações sugeridas:**

**Boa nota e muitas avaliações:** Considerar destaque na campanha
**Boa nota e poucas avaliações:** Testar maior exposição, considerando a menor evidência disponível
**Nota baixa e muitas avaliações:** Investigar reclamações antes de ampliar a promoção
**Desconto elevado e nota baixa:** Verificar problemas de experiência e revisar a estratégia comercial
